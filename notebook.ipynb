## Things to keep in mind

### Corrupt files

The Dataverse does not care if you upload corrupt files or not, and does not check files for any corruption. When you upload files to the Dataverse it would be good to know if the file is corrupt before you upload it. But how do we check for corrupt files? Unfortunately, many files such as zip files require special methods that are specific to that file type be called to check the file for corruption. Just know that this Notebook does not check for corrupt files. **Note: If a single zip file is uploaded to the Dataverse and it DOES NOT extract, then it is likely corrupt.**
