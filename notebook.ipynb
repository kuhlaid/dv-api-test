# Working with the Dataverse

The following are the methods for working with the Dataverse (e.g. creating collections, creating datasets, uploading files, updating the dataset version with new files, etc.). Again, everything that is personalized is found in the `_config` file, so most of the strings being passed to each method below (e.g. "objDvApi_COLLECTION_START") are telling the Python code to use the property from the `_config` file.

```python
from _worker import Worker
objWorker = Worker("_config.json") # initialize our Worker object (**MUST DO**)
```

```python
objWorker.createCollection("objDvApi_COLLECTION_START")
```

```python
objWorker.viewCollection("objDvApi_COLLECTION_START") # helper to verify a collection was created
```

```python
objWorker.createDataset("objDvApi_COLLECTION_START", "objDvApi_DATASET_INIT_PART")  # create a dataset for a collection
```

```python
objWorker.createZipFile("objMyZipFile") # create a zip archive
```

```python
objWorker.doubleZip("objMyZipFile") # double zips an archive
```

```python
objWorker.createTextFiles("lstTEST_FILES") # initial list of files to upload
```

```python
objWorker.uploadFiles("lstTEST_FILES") # send files to the dataset defined in the _config
```

```python
objWorker.updateDatasetMetadata("objDvApi_DATASET_UPDATE")  # update dataset metadata
```

```python
objWorker.publishDatasetDraft("objDvApi_COLLECTION_START", "major")
```

```python
objWorker.createTextFiles("lstTEST_FILES2")
```

```python
objWorker.createEmptyDatasetDraft() # will empty the draft dataset defined in `_cc__DvDatasetMetadata.json`
```

```python
objWorker.uploadFiles("lstTEST_FILES2")
```

```python
objWorker.viewDatasetFiles(":draft")  # show draft dataset contents
```

```python
objWorker.deleteDataset() # will delete the dataset defined in `_cc__DvDatasetMetadata.json`
```

```python
objWorker.deleteCollection("objDvApi_COLLECTION_START")  # delete our dataverse collection
```

```python
objWorker.viewCollectionContents("objDvApi_COLLECTION_START")  # view dataverse collection contents
```
